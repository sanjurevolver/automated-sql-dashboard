# Banking operations exploration
Run `pipeline.py` first. This notebook reads the latest successful run and does not mutate the warehouse. Use the project's Python environment as the kernel. All amounts are nominal CZK; ledger volume is not revenue.

```python
from pathlib import Path
import json
import pandas as pd
import plotly.express as px
root = Path.cwd()
if root.name == 'notebooks': root = root.parent
latest = json.loads((root / 'reports/latest.json').read_text())
report = Path(latest['report_directory'])
monthly = pd.read_csv(report / 'monthly_cashflow.csv', parse_dates=['month'])
accounts = pd.read_csv(report / 'account_performance.csv')
monthly.head()
```

```python
px.line(monthly, x='month', y=['inflow','outflow','net_flow'], title='Monthly ledger cash flows (CZK)')
```

```python
accounts[['transactions','gross_volume','average_amount','net_flow']].describe(percentiles=[.01,.5,.95,.99])
```

```python
px.histogram(accounts, x='average_amount', nbins=60, title='Account mean transaction amount distribution (CZK)')
```

```python
px.imshow(accounts[['transactions','gross_volume','average_amount','net_flow']].corr(method='spearman'), text_auto='.2f', title='Descriptive account-level Spearman correlation')
```
